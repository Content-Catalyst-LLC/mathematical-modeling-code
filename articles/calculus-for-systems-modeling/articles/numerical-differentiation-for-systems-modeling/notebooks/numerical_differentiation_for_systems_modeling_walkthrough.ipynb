# Numerical Differentiation for Systems Modeling

Placeholder for finite differences, forward/backward/central difference formulas, truncation error, roundoff error, noise amplification, step-size sensitivity, second derivatives, spatial gradients, and derivative governance.

$$f'(t)\approx \frac{f(t+h)-f(t)}{h}$$
$$f'(t)\approx \frac{f(t+h)-f(t-h)}{2h}$$
$$f''(t)\approx \frac{f(t+h)-2f(t)+f(t-h)}{h^2}$$
